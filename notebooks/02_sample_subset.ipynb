 Step 3: Take random 1k subset from combined_covid_vaers_relevant.csv

```python
import pandas as pd
import re
import os

# Define paths relative to root (go one level up from notebooks/)
SRC = os.path.join("..", "data", "processed", "combined_covid_vaers_relevant.csv")
DST = os.path.join("..", "data", "processed", "sample_1k.csv")

# Read CSV
df = pd.read_csv(SRC, low_memory=False)

# Clean VAX_NAME → extract text inside parentheses if present
df["VAX_NAME"] = df["VAX_NAME"].apply(
    lambda x: re.sub(r".*?\(([^)]*)\)?", r"\1", str(x)).strip() if "(" in str(x) else x
)

# Save random 1k sample
df.sample(1_000, random_state=42).to_csv(DST, index=False)

print(f"✅ Saved cleaned 1k sample to {DST}")
```
